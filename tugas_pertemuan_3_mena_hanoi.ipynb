```python
def tower_of_hanoi(n, asal, bantu, tujuan, langkah):
    """
    Fungsi rekursif untuk menyelesaikan Tower of Hanoi.
    n      : jumlah disk
    asal   : tiang asal (misal 'A')
    bantu  : tiang bantu (misal 'B')
    tujuan : tiang tujuan (misal 'C')
    langkah: list untuk menghitung total langkah
    """
    if n == 1:
        langkah[0] += 1
        print(f"Langkah {langkah[0]}: Pindahkan disk 1 dari {asal} ke {tujuan}")
        return

    # Pindahkan n-1 disk dari asal ke bantu
    tower_of_hanoi(n-1, asal, tujuan, bantu, langkah)

    # Pindahkan disk ke-n dari asal ke tujuan
    langkah[0] += 1
    print(f"Langkah {langkah[0]}: Pindahkan disk {n} dari {asal} ke {tujuan}")

    # Pindahkan n-1 disk dari bantu ke tujuan
    tower_of_hanoi(n-1, bantu, asal, tujuan, langkah)

# Skenario sesuai soal gambar: 4 disk dari A ke C
jumlah_disk = 4
penghitung_langkah = [0]

print(f"Solusi Menara Hanoi untuk {jumlah_disk} disk:")
print("-" * 40)
tower_of_hanoi(jumlah_disk, 'A', 'B', 'C', penghitung_langkah)
print("-" * 40)
print(f"Total langkah minimum yang diperlukan: {penghitung_langkah[0]}")
```
